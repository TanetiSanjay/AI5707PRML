```python
import os
import requests
from PIL import Image
from io import BytesIO
import time

# ----------------------------
# Configuration
# ----------------------------
SAVE_DIR = "AAH"
NUM_IMAGES = 22
SIZE = (512, 512)

os.makedirs(SAVE_DIR, exist_ok=True)

saved = 0
attempt = 0

while saved < NUM_IMAGES:
    url = f"https://cataas.com/cat?random={attempt}"

    try:
        print(f"Attempt {attempt} (saved {saved}/{NUM_IMAGES})")

        response = requests.get(
            url,
            headers={"User-Agent": "Mozilla/5.0"},
            timeout=15,
        )
        response.raise_for_status()

        img = Image.open(BytesIO(response.content))

        # Skip square images
        if img.width == img.height:
            print(f"  Skipping square image ({img.width}×{img.height})")
            attempt += 1
            continue

        # Resize
        # img = img.resize(SIZE, Image.LANCZOS)

        img.save(os.path.join(SAVE_DIR, f"cat_{saved+1:02d}.png"))

        saved += 1
        attempt += 1
        time.sleep(0.5)

    except Exception as e:
        print(f"Failed on attempt {attempt}: {e}")
        attempt += 1

print("Done!")
```
